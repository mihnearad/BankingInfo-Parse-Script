```python
import pandas as pd

df_rev=pd.read_csv('/home/mihnearad/Development/BankingInfo-Parse-Script/.data/REVBUS.csv')
```

```python
df_rev.head()
```

```python
new_order_column=['Date started (UTC)','Description','Reference','Amount','Fee currency']

df_rev=df_rev[new_order_column]
df_rev.head()
```

```python
df_rev["Date started (UTC)"] = (
    pd.to_datetime(df_rev["Date started (UTC)"])
    .dt.strftime("%d-%m-%Y")
    .sort_values(ascending=True)
)
```

```python
filtered_rev = df_rev[~df_rev["Description"].str.contains(r"Mihnea|RON")]
filtered_rev 
```

```python
filtered_rev = filtered_rev.rename(columns={
    'Date started (UTC)': 'Payment Date',
    'Description': 'Merchant',
    'Reference': 'Description',
    'Amount': 'Debit',
    'Fee currency': 'Currency'
})

filtered_rev["Account"]= "Rev Business"
filtered_rev['Payment Date'] = pd.to_datetime(filtered_rev['Payment Date'], format='%d-%m-%Y')



filtered_rev.head()
```

```python

filtered_rev.to_csv('/home/mihnearad/Development/BankingInfo-Parse-Script/.data/filtered/filtered_revbus.csv',index=False)
```
